### Django Project Structure (after `django-admin startproject nameofproject`)

```
nameofproject/
│── manage.py      → project manager (runserver, migrate, etc.)
│
└── core/          → main project folder
    │── __init__.py → marks folder as Python package
    │── asgi.py     → entry for ASGI server (async)
    │── settings.py → project config (DB, apps, middleware)
    │── urls.py     → main URL routes
    │── wsgi.py     → entry for WSGI server (deploy)
```

---

### Quick Meaning

* `manage.py` → controls project commands
* `settings.py` → config/brain of project
* `urls.py` → URL mapping
* `wsgi.py` / `asgi.py` → server entry points
